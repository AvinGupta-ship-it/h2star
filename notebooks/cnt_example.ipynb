# CNT example (documented notebook example; NOT a paper claim)

The pre-registered rule (10.5281/zenodo.23931242 section 8) required eight or
more dual-digitization-reconciled points on a 77 K high-pressure CNT isotherm
for a CNT case study to appear in the paper. The only traceable candidate —
Panella, Hirscher & Roth, *Carbon* 43 (2005) 2209–2214, Fig. 2(b), purified
SWCNTs — reconciles to **five** points (`data/validation/panella_fig2b_swcnt.csv`),
so the case study was **dropped**; this notebook is the documented example the
rule prescribes instead. It makes no system-level claim.

What the five points show, on both measurement bases, is consistency with the
established literature result that carbon nanotubes do not exceed
high-surface-area activated carbon for hydrogen storage (Tibbetts et al. 2001;
Panella et al. 2005 itself; Zhou, Zhou & Sun 2004, who measured AX-21 and MWNTs
on the same apparatus).

On the record, two corrections to v0.2 wording: (1) Wang et al. 2002's 8.0 wt%
at ambient pressure is **reachable** in the modified D–A model, but only at a
limiting uptake far beyond the physically plausible bracket — the v0.2 claim
that "no limiting uptake at all" reproduces it was wrong and is withdrawn.
(2) The v0.2 F8 waterfall's first bar mixed inter-particle gas into a
"material uptake" label; that figure is withdrawn with the case study.


```python
import csv
import os

# Resolve the repository root so the notebook runs from the repo root or
# from notebooks/ alike.
ROOT = os.getcwd()
while not os.path.isdir(os.path.join(ROOT, 'data', 'validation')):
    parent = os.path.dirname(ROOT)
    if parent == ROOT:
        raise FileNotFoundError('repository root not found above cwd')
    ROOT = parent

rows = list(csv.DictReader(
    open(os.path.join(ROOT, 'data', 'validation', 'panella_fig2b_swcnt.csv'))))
cnt77 = [(float(r['pressure_bar']), float(r['wt_percent'])) for r in rows if r['temperature_K']=='77']
print('Panella SWCNT 77 K (reconciled):', cnt77)
for P, w in cnt77:
    print(f'  {P:5.1f} bar  {w:.2f} wt%')
```

The SWCNT plateau (≈2.4–2.5 wt% excess at 77 K) sits far below AX-21-class
carbons (≈5.4 wt% excess at 77 K per Bénard & Chahine 2001), broadly in line
with surface-area scaling (the Chahine rule; Goldsmith et al. 2013), with the
paper's own error bars (digitized here) spanning ±0.1–0.3 wt%. Any transfer of
these five points into system space would be dominated by parameter-transfer
assumptions, which is exactly why the pre-registered rule excluded it from the
paper.
